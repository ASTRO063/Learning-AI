### why we need agent?


```python
from langchain.chat_models import init_chat_model

local_model = init_chat_model(
    model="local-llm",
    base_url="http://localhost:8080/v1",
    api_key="dummy",
    model_provider="openai",
    temperature=0.7)
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model = local_model,
    system_prompt = "You are a specialised agent in software engineering"
)
```

```python
from langchain_core.messages import HumanMessage
response = agent.invoke(
    {"messages": [HumanMessage("who are you")] }
)
print(response)
```

```python
response["messages"][-1].content
```

```python
from pprint import pprint
def print_ai_reply(response):
    pprint(response["messages"][-1].content)
```

```python
response = agent.invoke(
    {"messages": [HumanMessage("what is capital city of Hyderabad")] }
)
print_ai_reply(response)
```

### Ways to parse or get response
- We have seen how to invoke and parse the response
- Is it only way to do it? --> No
- We can also stream into the response.

```python
## Here we will see one of the streaming options

for token, metadata in agent.stream(
    {"messages": [HumanMessage("what is capital city of Hyderabad")] },
    stream_mode = "messages"
):
    # print("token-->", token)
    # print("metadata--->", metadata)
    if token.content:
        print(token.content, end="", flush=True)
```
