## Integrating Local LLMs (Project Focus)

For our project, we need to connect LangChain to a model running locally (e.g., via `llama-server`). This allows us to work offline and experiment with models like Gemma.

### Setup for Local Model
1. Start the local server (from README): `llama-server -m ./models/<your_model>.gguf --port 8080`
2. Ensure the model file is in the correct path.

````python
from langchain_openai import ChatOpenAI

local_model_with_open_ai = ChatOpenAI(
    base_url="http://localhost:8080/v1",
    model="local-llm",
    api_key="dummy",
    temperature=0.7)

response = local_model_with_open_ai.invoke("who are you?")
print(response)
````

````python
from langchain.chat_models import init_chat_model

local_model_with_abstract = init_chat_model(
    model="local-llm",
    base_url="http://localhost:8080/v1",
    api_key="dummy",
    model_provider="openai",
    temperature=0.7)
response = local_model_with_abstract.invoke("who are you")
print(response)
````
