# Host llm locally or any other server


### Host local

Refer README.md file for running llm locally

````python
from langchain_openai import ChatOpenAI

local_model_with_open_ai = ChatOpenAI(
    base_url="http://localhost:8080/v1",
    model="local-llm",
    api_key="dummy",
    temperature=0.7)

response = local_model_with_open_ai.invoke("who are you?")
print(response)
````

````python
from langchain.chat_models import init_chat_model

local_model_with_abstract = init_chat_model(
    model="local-llm",
    base_url="http://localhost:8080/v1",
    api_key="dummy",
    model_provider="openai",
    temperature=0.7)
response = local_model_with_abstract.invoke("who are you")
print(response)
````

Use ChatOpenAI when your app is strictly tied to OpenAI or an OpenAI proxy and you want strict class-level type completions. Use init_chat_model when you want an abstract initializer that lets you switch between different model providers without refactoring.
